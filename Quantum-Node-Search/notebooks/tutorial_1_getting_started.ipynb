# Quantum Search in Graph Nodes - Tutorial 1: Getting Started

This notebook introduces the basics of the Quantum Search project and how to use it.

## Prerequisites
- Python 3.8+
- Installed dependencies: `pip install -r requirements.txt`
- Basic understanding of graphs and search algorithms

## 1. Simple Graph Creation and Visualization

```python3
# Import the graph generation module
from graphs import create_graph, GraphGenerator
import matplotlib.pyplot as plt
import networkx as nx

# Create a simple random graph with 16 nodes
graph = create_graph('random', n_nodes=16, seed=42)

print(f"Graph created with {graph.number_of_nodes()} nodes and {graph.number_of_edges()} edges")

# Get graph statistics
generator = GraphGenerator()
stats = generator.get_graph_stats(graph)

print("\nGraph Statistics:")
for key, value in stats.items():
    print(f"  {key}: {value}")
```

## 2. Visualizing the Graph

```python3
# Visualize the graph
plt.figure(figsize=(10, 8))

# Use spring layout for better visualization
pos = nx.spring_layout(graph, k=0.5, iterations=50, seed=42)

# Draw the network
nx.draw_networkx_nodes(graph, pos, node_color='lightblue', node_size=500)
nx.draw_networkx_edges(graph, pos, alpha=0.3)
nx.draw_networkx_labels(graph, pos, font_size=10)

plt.title("Random Graph (16 nodes)")
plt.axis('off')
plt.tight_layout()
plt.show()
```

## 3. Classical Search - Linear

```python3
from classical import run_classical_search

# Define target node
target = 7

# Run linear search
result = run_classical_search(graph, target, method='linear')

print(f"Target Node: {target}")
print(f"Found: {result['found']}")
print(f"Nodes Checked: {result['nodes_checked']}")
print(f"Execution Time: {result['execution_time']*1000:.4f} ms")
print(f"Complexity: {result['complexity']}")
print(f"Search Path: {result['path'][:10]}...")  # Show first 10 nodes
```

## 4. Classical Search - BFS

```python3
# Run BFS search
result_bfs = run_classical_search(graph, target, method='bfs')

print(f"BFS Results for Target {target}:")
print(f"Found: {result_bfs['found']}")
print(f"Nodes Checked: {result_bfs['nodes_checked']}")
print(f"Execution Time: {result_bfs['execution_time']*1000:.4f} ms")
print(f"Search Path (length {len(result_bfs['path'])}): {result_bfs['path']}")
```

## 5. Quantum Search - Grover's Algorithm

```python3
from quantum import run_grover_search
import math

# Run Grover's algorithm
# Note: For 16 nodes, we need 4 qubits
result_grover = run_grover_search(n_nodes=16, target=target, shots=1000)

print(f"Grover's Algorithm Results for Target {target}:")
print(f"Found: {result_grover['found']}")
print(f"Measured Node: {result_grover['measured_node']}")
print(f"Grover Iterations: {result_grover['grover_iterations']}")
print(f"Execution Time: {result_grover['execution_time']*1000:.4f} ms")
print(f"Success Probability: {result_grover['success_probability']:.1f}%")
print(f"\nTheoretical Iterations: k ≈ π/4 × √N = π/4 × √16 = {math.pi/4 * math.sqrt(16):.1f}")
print(f"Complexity: {result_grover['complexity']}")
```

## 6. Comparison: Classical vs Quantum

```python3
import pandas as pd

# Create comparison dataframe
comparison = pd.DataFrame({
    'Algorithm': ['Linear', 'BFS', 'Grover'],
    'Steps': [
        result['nodes_checked'],
        result_bfs['nodes_checked'],
        result_grover['grover_iterations']
    ],
    'Time (ms)': [
        result['execution_time'] * 1000,
        result_bfs['execution_time'] * 1000,
        result_grover['execution_time'] * 1000
    ],
    'Complexity': [
        'O(N)',
        'O(N+E)',
        'O(√N)'
    ]
})

print(comparison)

# Calculate speedup
speedup = result['nodes_checked'] / result_grover['grover_iterations']
print(f"\nQuantum Speedup: {speedup:.2f}x")
print(f"Expected Speedup: √16 = {math.sqrt(16):.1f}x")
```

## 7. Visualizing Comparison

```python3
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Steps comparison
axes[0].bar(comparison['Algorithm'], comparison['Steps'], color=['#3498db', '#95a5a6', '#e74c3c'])
axes[0].set_ylabel('Number of Steps/Iterations')
axes[0].set_title('Algorithm Steps Comparison')
axes[0].grid(axis='y', alpha=0.3)

# Time comparison
axes[1].bar(comparison['Algorithm'], comparison['Time (ms)'], color=['#3498db', '#95a5a6', '#e74c3c'])
axes[1].set_ylabel('Execution Time (ms)')
axes[1].set_title('Execution Time Comparison')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 8. Key Takeaways

1. **Classical vs Quantum**: Quantum search provides O(√N) speedup
2. **Grover Iterations**: Always ≈ π/4 × √N
3. **Practical Advantage**: For large N, quantum is significantly faster
4. **Success Probability**: Quantum needs multiple shots but high success rate
5. **Graph Independence**: Works with any graph structure
